The resulting DataFrame should look conceptually like:

|---------|------------|-----------|------------|

```python
def compare_students(students_records, student1, student2):

    rows=[]
    for student in students_records:
        student_name=student['Name']
        for subject_key, subject_scores in student['Assessments'].items():
            for index, row in subject_scores.iterrows():
                Assignment=row['Assignment']
                Score=row['Score']
                rows.append({'Name':student_name,'Subject':subject_key,'Assignment':Assignment,'Score':Score})

    df_all=pd.DataFrame(rows)   
    student1_subjects=set(df_all[df_all['Name']==student1]['Subject'].unique())
    student2_subjects=set(df_all[df_all['Name']==student2]['Subject'].unique())

    common_subjects= list(student1_subjects & student2_subjects)

    df_average=df_all.groupby(['Name','Subject'])['Score'].mean().reset_index()
    df_average=df_average[(df_average['Subject'].isin(common_subjects))&(df_average['Name'].isin([student1, student2]))].reset_index(drop=True)
    
    sumary_table= df_average.pivot(
        index='Subject',
        columns='Name',
        values='Score'
    ).reset_index()
    sumary_table = sumary_table[['Subject', student1, student2]]

    sumary_table= (sumary_table.rename(
        columns={
            student1:student1+'_averaged',
            student2:student2+'_averaged'
        }
    ))
    sumary_table['Difference']=sumary_table[student1+'_averaged']-sumary_table[student2+'_averaged']


    return sumary_table

sumary_table= compare_students(students_records,'Liyuan','Alice')
display (sumary_table)

print(sumary_table.columns)
print(sumary_table.index)
```
