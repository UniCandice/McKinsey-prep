```python

from sklearn.ensemble import RandomForestRegressor 
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score

df_model=df_time_series.sort_values('Test_Time').reset_index(drop=True).copy()


df_model=df_model[['Test_Time','Voltage','Current','Cycle_Index']]
df_model.info()
```

```python
df_time_series.Test_Time.is_monotonic_increasing and df_time_series.Test_Time.is_unique
```

```python
df_time_series[df_time_series['Test_Time'].diff()<0]
```

```python
lag_steps = [1,2,3,5,10]

for lag in lag_steps:
    df_model[f'Voltage_lag_{lag}']=df_model['Voltage'].shift(lag)

df_model.info()
df_model.head()
df_model.isna().any(axis=1)
df_model.isna().any(axis=0)

    
```

```python
nan_rows = df_model[df_model.isna().any(axis=1)]
nan_rows.isna()
```

```python
rolling_windows = [3,5,10]

for window in rolling_windows:
    df_model[f'Voltage_roll_mean_{window}']=df_model['Voltage'].shift(1).rolling(window).mean()
    df_model[f'Voltage_roll_std_{window}']=df_model['Voltage'].shift(1).rolling(window).std()


df_model['Voltage_diff']=df_model['Voltage'].diff(1)
df_model['Current_lag_1']=df_model['Current'].shift(1)
df_model['Time_delta']=df_model['Test_Time'].diff(1)

feature_columns = (
    [f'Voltage_lag_{lag}' for lag in lag_steps]
    +[f'Voltage_roll_mean_{window}' for window in rolling_windows]
    +[f'Voltage_roll_std_{window}' for window in rolling_windows]
    +['Voltage_diff','Current_lag_1','Time_delta','Current', 'Cycle_Index']
)

df_model.info()

type(feature_columns)
print(feature_columns)
```

```python
df_model.info()
```

```python
df_model.isna().any(axis=0)
```

```python
df_model = df_model.dropna(subset=feature_columns+['Voltage']).reset_index(drop=True)
```

```python
df_model.isna().any(axis=0)
```

```python
split_index = int (len(df_model)*0.8)
print(f'split_index value is {split_index} and the length of the df_model is {len(df_model)}')

train_data = df_model.iloc[:split_index]
test_data = df_model.iloc[split_index:]

x_train, y_train = train_data[feature_columns],train_data['Voltage']
x_test,y_test =test_data[feature_columns],test_data['Voltage']
```

```python
model=RandomForestRegressor(n_estimators=300,max_depth=12,random_state=42,n_jobs=-1)
model.fit(x_train,y_train)
```

```python
predicted_voltage=model.predict(x_test)
```

```python
mean_absolute_error_score=mean_absolute_error(y_test,predicted_voltage)
rmse_score=np.sqrt(mean_squared_error(y_test,predicted_voltage))
r_squared_score=r2_score(y_test,predicted_voltage)

print(f'MAE: {mean_absolute_error_score}')
print(f'RMSE: {rmse_score}')
print(f'R2: {r_squared_score}')
```

```python
# actual vs predicted plot 

plt.figure()
plt.plot(test_data.Test_Time, test_data.Voltage)
plt.plot(test_data.Test_Time,predicted_voltage)
plt.xlabel('test time ')
plt.ylabel('Voltage')
plt.legend(['actual voltage','predicted voltage'])
plt.show()
```
